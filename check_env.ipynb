# بررسی محیط پروژه پردازش صوت

این نوت‌بوک نسخه پایتون، مسیر اجرای پایتون و نسخه کتابخانه‌های مورد استفاده را ثبت می‌کند. آن را در همان محیطی اجرا کنید که نوت‌بوک پروژه را اجرا خواهید کرد. سپس فایل را ذخیره کنید تا خروجی‌ها برای تحویل باقی بمانند.

```python
import importlib.metadata as metadata
import platform
import sys

print("Python version:", sys.version)
print("Python executable:", sys.executable)
print("Platform:", platform.platform())

packages = {
    "NumPy": "numpy",
    "Matplotlib": "matplotlib",
    "pandas": "pandas",
    "librosa": "librosa",
    "scikit-learn": "scikit-learn",
}
for label, distribution in packages.items():
    try:
        version = metadata.version(distribution)
    except metadata.PackageNotFoundError:
        version = "NOT INSTALLED"
    print(f"{label}: {version}")
```
